# Ejercicio: Encuentre la probabilidad de que el movimiento browniano con parámetro deriva, $\mu=0.6$ y varianza $\sigma^2=0.25$ tome valores entre 1 y 3 al tiempo 4.

# Solución analítica 
El movimiento browniano con deriva y con los parámetros dados es: $x_t=0.6t+0.5B_t$. Sustituyendo $t=4$<br>
$P(1<x_t<3)=P(1<2.4+0.5B_4<3)=P(-1.4<0.5B_4<0.6)=P(-2.8<B_4<1.2)$<br>
Pero $B_t$ sigue una distribución normal con media $0$ y varianza $t=4$. Así:<br>
$P(-2.8<B_4<1.2)=\int_{-2.8}^{1.2}\frac{1}{\sqrt{2*4*\pi}} e^{-\frac{x^2}{8}}dx=0.645$

# Solución por simulación

```python
import matplotlib.pyplot as plt
import numpy as np
from math import sqrt


N = 1000
N1=10000
M=0
XM=np.zeros(N)
XM1=[]
for j in range(N):
  t = np.linspace(0, 4, N1)
  dt = 4 / N1
  w=np.zeros(N1)
  x=np.zeros(N1)
  for i in range(1,N1):
    w[i]=w[i-1]+sqrt(dt)*np.random.normal(0,1)
    x[i]=0.6*t[i]+0.5*w[i]
  if x[N1-1]>=1 and x[N1-1]<=3:
    M+=1
    XM1.append(x[N1-1])
  XM[j]=x[N1-1]
plt.figure()
plt.hist(XM,bins=50,label='Caminatas que no cumplen la condición')
plt.hist(XM1,bins=15,color='purple', label='Caminatas que si cumplen la condición')
plt.show()
print('Probabilidad de que 1<= x_4 <= 3 =',M/N) 
```
