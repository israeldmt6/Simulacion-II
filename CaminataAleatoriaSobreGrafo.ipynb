## Caminata aleatoria sobre un grafo

# Caminata aleatoria sobre un grafo para estimar el número esperado de $1's$ en una buena cadena.
Considérese una secuencia de unos y ceros de longitud n, llamaremos una buena cadena a la que no contenga unos adyacentes. Por ejemplo para $n=4$ las buenas cadenas son:

$(0000)$  $(1000)$  $(0100)$  $(0010)$  $(0001)$  $(1010)$ $(0101)$ $(1001)$

La pregunta a responder es ¿Cuál es la cantidad de $1's$ esperados en una buena cadena?
Una forma de resolverlo es a partir de la siguiente ecuación:
                                $\mu=\sum_{k}^{}k\hspace{1mm}\pi_k$
donde $\pi_k$ es al probabilidad e que una buena cadena tengua $k$ unos.
Así para obtener este valor se deberían generar todas las cadenas posibles y luego elegir solo las buenas. Sin embargo esto puede ser complicado para cadenas de longitudes grandes, pues por ejemplo para cadenas de longitud 100 la probalidad de obtener una buena cadenaes cercana a $10^{-9}$. A continuación se muestra un programa que genera 10 cadenas de longitud $n=5$ de forma aleatoria y se nombran buenas o malas cadenas. Puede verse que la cantidad de buenas cadenas es menor que las malas cadenas.


```python
from random import random
n=5
for j in range(10):
  Y=''
  for i in range(n):
      u=random()
      if u<0.5: Y+='0'
      else: Y+='1'
  if '11' not in Y: print(Y, "Buena cadena ")
  else: print(Y, "Mala cadena")
```

Por ello para estimar el valor solicitado, se procede de la siguiente forma.<br>
Si $x$ es una cadena de longitud $n$. Sea $r(x)$ el número de $1's$ de la cadena. <br>
Si nos restringimos a solamente buenas cadenas  $Y_1, Y_2, ... Y_k$ podemos estimar la respuesta por método de montecarlo: <br>
                                $\mu=\frac{r(Y_1)+r(Y_2)+...+r(Y_k)}{k}$
<br>                            
Entonces para generar solamente buenas cadenas se usa un método de aceptación y rechazo, dada una buena cadena se elige uno de sus $n$ componentes de forma aleatoria. <br>
<ul>
    <li>1. Si la componente es 1 entonces se cambia a 0. Dado que el número de 1 es reducido el resultado seguirá siendo una buena cadena. </li>
    <li>2. Si la componente es 0 entonces se cambia a 1 solo si el resultado es una buena cadena. Así se obtiene una buena cadena. Sin embargo, si el cambio de 0 a 1 no genera una buena cadena entonces se conserva la cadena actual sin cambios.</li>
</ul>
A continuación se muestra el programa que estima la cantidad de $1's$ para cadenas de longitud $n=100$
    

Se cargan las librerías a usar, random para elegir una componente aleatoria de la cadena, statistics para calcular la media de un conjunto de datos y numpy porque se usarán arreglos tipo np

```python
import random
from statistics import mean
import numpy as np
```

Se genera la primera buena cadena que tiene en todas sus componentes 0 excepto en el último, y se usará una lista para guardar la cantidad de $1's$ en cada buena cadena generada.

```python
n = 100
Y = np.zeros(n,dtype=int)
Y[n-1]=1
l = []
```

Se generarán 1000000 buenas cadenas, primero se elige la componente a cambiar, pero se usará una cadena auxiliar que será una copia de la originial, esta cadena será la que se le configure su componente previamente elegido. Luego de ser alterado se comprueba si es una buena cadena, para ello se recorren todas sus componente buscando unos adyacentes, si no los hay entonces la cadena original será igual a la cadena auxiliar, de no ser así se queda igual, por último la cantidad de unos que tiene la cadena se guarda en la lista.

```python
for i in range(1000000):
  # Se elige la componente a configurar  
  u = random.randint(0,n-1)
  # Se hace una copia  
  aux = Y.copy() 
  # Se configura la componente elegida de la cadena auxiliar  
  if Y[u] == 1:
      aux[u] = 0
  else:
      aux[u] = 1
  # Se inicializa una variable booleana que guardará si la cadena auxiliar es buena cadena o no 
  a = True
  # Se recorre toda la cadena buscando 1's aduyacentes  
  for j in range(n-1):
    if aux[j] == 1 and aux[j+1] == 1:
      a = False
      break
  # Si no se encontraron 1's adyacentes la cadena es buena y entonces la cadena original será igual a la auxliar      
  if a:
    Y = aux
  # Si la cadena auxiliar no es buena cadena entonces la cadena original no cambia
  # Se guarda en la lista la cantidda de 1's que tiene la cadena
  l.append(np.count_nonzero(Y==1))
```

El valor esperado de $1's$ en una cadena buena de longitud $100$ es:

```python
print(mean(l))
```
